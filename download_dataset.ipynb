```python
# I will create a dataset only for customers table
df_customers = df[["customer_id", "first_name", "last_name",'gender', 'age_group', 'signup_date', 'country']].drop_duplicates().reset_index(drop=True).to_csv("customers.csv", index=False)
```

```python
# I will create a dataset only for products table
df_products = df[["product_id", "product_name", "category", "unit_price", ]].drop_duplicates().reset_index(drop=True).to_csv("products.csv", index=False)
```

```python
# I will create a dataset only for orders table
df_orders = df[["order_id", "customer_id", "order_date", "order_status", "payment_method"]].drop_duplicates().reset_index(drop=True).to_csv("orders.csv", index=False)
```

```python
# Select columns you want (excluding order_item_id, since we’ll create it)
df_order_items = df[["order_id", "product_id", "quantity", "unit_price"]].drop_duplicates().reset_index(drop=True)

# Create ascending order_item_id starting from 1
df_order_items["order_item_id"] = range(1, len(df_order_items) + 1)

# Reorder columns
df_order_items = df_order_items[["order_item_id", "order_id", "product_id", "quantity", "unit_price"]]

# Save to CSV
df_order_items.to_csv("order_items.csv", index=False)
```

```python
# I will create a dataset only for reviews table
df_reviews = df[["review_id", "product_id", "customer_id", "rating", "review_text", "review_date"]].drop_duplicates().reset_index(drop=True).to_csv("reviews.csv", index=False)
```
